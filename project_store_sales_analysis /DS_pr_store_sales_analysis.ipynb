````python
# Запусти цей блок, щоб отримати датасет
import pandas as pd
import numpy as np

np.random.seed(42)
n = 300

categories = ['Молочне', 'Хліб та випічка', 'М\'ясо та риба', 'Овочі та фрукти', 'Напої']
products = {
    'Молочне':          ['Молоко 1л', 'Кефір 0.5л', 'Сир кисломолочний', 'Масло вершкове', 'Йогурт'],
    'Хліб та випічка':  ['Хліб білий', 'Батон', 'Булочка', 'Хліб житній', 'Круасан'],
    'М\'ясо та риба':   ['Куряче філе', 'Свинина', 'Ковбаса', 'Риба хек', 'Сосиски'],
    'Овочі та фрукти':  ['Картопля', 'Морква', 'Яблука', 'Банани', 'Помідори'],
    'Напої':            ['Вода 1.5л', 'Сік апельсиновий', 'Чай', 'Кава мелена', 'Лимонад'],
}
prices = {
    'Молоко 1л': 38, 'Кефір 0.5л': 22, 'Сир кисломолочний': 55, 'Масло вершкове': 95, 'Йогурт': 30,
    'Хліб білий': 28, 'Батон': 25, 'Булочка': 12, 'Хліб житній': 32, 'Круасан': 18,
    'Куряче філе': 165, 'Свинина': 210, 'Ковбаса': 145, 'Риба хек': 120, 'Сосиски': 98,
    'Картопля': 20, 'Морква': 15, 'Яблука': 45, 'Банани': 52, 'Помідори': 60,
    'Вода 1.5л': 18, 'Сік апельсиновий': 68, 'Чай': 85, 'Кава мелена': 155, 'Лимонад': 35,
}

chosen_cats = np.random.choice(categories, n, p=[0.25, 0.20, 0.20, 0.20, 0.15])
chosen_products = [np.random.choice(products[c]) for c in chosen_cats]

df_origin = pd.DataFrame({
    'transaction_id': range(1001, 1001 + n),
    'date':           pd.to_datetime(
                          np.random.choice(pd.date_range('2024-03-01', '2024-03-31'), n)
                      ),
    'hour':           np.random.choice(range(8, 21), n, p=[0.04,0.06,0.08,0.10,0.12,0.12,0.11,0.10,0.09,0.08,0.06,0.04,0.00]),
    'category':       chosen_cats,
    'product':        chosen_products,
    'quantity':       np.random.randint(1, 6, n),
    'price_uah':      [prices[p] for p in chosen_products],
    'customer_age':   np.random.randint(18, 70, n),
    'payment':        np.random.choice(['Готівка', 'Картка', 'Телефон'], n, p=[0.35, 0.50, 0.15]),
})
````

## Блок 4 - Нові стовпці та трансформація

1. Додай стовпець revenue_share - частка виторгу кожної транзакції від загального виторгу (у відсотках).

````python
total_revenue = df.total_uah.sum()
df['revenue_share'] = (df.total_uah / total_revenue * 100).round(4)

df[['product', 'total_uah', 'revenue_share']].head()
````

2. Додай стовпець age_group:  
  'Молодь' - до 30 років  
  'Середній вік' - 30–50 років  
  'Старші' - понад 50 років

````python
df['age_group'] = pd.cut(
    df['customer_age'],
    bins=[0, 30, 50, 100],
    labels=['Молодь', 'Середній вік', 'Старші'],
    right=False  # [0, 30), [30, 50), [50, 100)
)

print("Розподіл по віковим групам:\n")
# df.head()
df.age_group.value_counts()
````

3. Додай стовпець weekday - день тижня з колонки date.

4. Додай булевий стовпець is_big_purchase - True, якщо total_uah > 300.

````python
df['is_big_purchase'] = df.total_uah > 300
pct = df['is_big_purchase'].mean() * 100
print(f'Частка \'великих\' покупок (> 300): {pct:.1f}%')

# 1, 0, 0, 0, 1, 1, 1 -> 7
# 1 -> 4, 3 -> 0
# 1 від загального -> 4 / 7 * 100%
````

````python
# Розподіл по віковим групам
print("Середній чек по віковим групам:\n")
df.groupby('age_group', observed=True)['total_uah'].mean().round(2)
````

````python
# Розподіл по віковим групам
print("Середній чек по віковим групам і кількість 'великих' покупок:\n")
df.groupby('age_group', observed=True).agg({
    'total_uah': "mean",
    'is_big_purchase': 'sum'
}).round(2)
````

Напиши у вигляді коментарів у коді (або окремо) відповіді на питання власника магазину:

1.Яка категорія товарів найприбутковіша?


````python
rev_cat = df.groupby('category')['total_uah'].sum().sort_values(ascending=False)
top_cat = rev_cat.index[0]
top_val = rev_cat.iloc[0]
share   = top_val / rev_cat.sum() * 100

print("1. НАЙПРИБУТКОВІША КАТЕГОРІЯ")
print("─" * 50)
print(rev_cat.apply(lambda x: f"{x:>8,.0f} грн  ({x/rev_cat.sum()*100:.1f}%)").to_string())
print(f"\n→ Лідер: «{top_cat}» — {top_val:,.0f} грн ({share:.1f}% від загального виторгу)")
print(f"  Причина: найвищі ціни на одиницю товару (Свинина — 210 грн, Куряче філе — 165 грн)\n")
````

2.В які години найактивніша торгівля?


````python
hour_counts = df['hour'].value_counts().sort_index()
peak_hour   = hour_counts.idxmax()
peak_count  = hour_counts.max()

print("2. ПІКОВІ ГОДИНИ ТОРГІВЛІ")
print("─" * 50)
print(hour_counts.to_string())
print(f"\n→ Пік: {peak_hour}:00 — {peak_count} транзакцій")
print(f"  Активна торгівля: з 11:00 до 16:00 (обідня та після-обідня зона)")
print(f"  Мінімум: до 10:00 і після 19:00\n")
````

3.Який спосіб оплати переважає?


````python
payment_pct = df['payment'].value_counts(normalize=True) * 100
top_payment = payment_pct.index[0]

print("3. СПОСІБ ОПЛАТИ")
print("─" * 50)
print(payment_pct.apply(lambda x: f"{x:.1f}%").to_string())
print(f"\n→ Переважає: {top_payment} ({payment_pct.iloc[0]:.1f}%)")
print(f"  Середній чек по оплаті:")
print(df.groupby('payment')['total_uah'].mean().round(2).apply(lambda x: f"  {x:.2f} грн").to_string())
print()
````

4.Чи є «пікові» дні продажів?


````python
daily_rev    = df.groupby('day')['total_uah'].sum()
best_day_n   = daily_rev.idxmax()
worst_day_n  = daily_rev.idxmin()
mean_day_rev = daily_rev.mean()
threshold    = mean_day_rev * 1.5  # пік = більше ніж 150% від середнього
peak_days    = daily_rev[daily_rev > threshold]

print("4. ПІКОВІ ДНІ ПРОДАЖІВ")
print("─" * 50)
print(f"  Середній щоденний виторг: {mean_day_rev:,.0f} грн")
print(f"  Найкращий день: {best_day_n} березня — {daily_rev[best_day_n]:,.0f} грн")
print(f"  Найгірший день: {worst_day_n} березня — {daily_rev[worst_day_n]:,.0f} грн")
print(f"\n  Пікових днів (>150% від середнього): {len(peak_days)}")
if len(peak_days) > 0:
    for day, val in peak_days.items():
        print(f"    → {day} березня: {val:,.0f} грн")
print()
````
