**Нагадування:** факторіал — це добуток всіх чисел від 1 до `n`.

> 💡 **Підказка:** Створіть змінну `result = 1`, а потім у циклі множте її на кожне число.

### Завдання 27. Прості числа ⭐⭐⭐

Виведіть всі прості числа від 2 до 50.

**Нагадування:** просте число — це число, яке ділиться тільки на 1 і саме на себе.

> 💡 **Підказка:** Для кожного числа `n` від 2 до 50 перевіряйте, чи має воно дільники від 2 до n-1. Можна використати змінну-прапорець `is_prime = True`.

````python
# Ваш код для завдання 27:
for n in range(2, 51):          # перебираємо числа від 2 до 50
    is_prime = True             # спочатку вважаємо просте

    for i in range(2, n):       # перевіряємо дільники від 2 до n-1
        if n % i == 0:          # знайшли дільник!
            is_prime = False    # число НЕ просте
            break               # більше перевіряти не треба

    if is_prime:                # після внутрішнього циклу
        print(n)
````

> 💡 **Підказка:** У циклі: беремо останню цифру через `n % 10`, додаємо до суми, потім відкидаємо її через `n = n // 10`. Цикл працює, поки `n > 0`.

### Завдання 56. Сортування без `.sort()` ⭐⭐⭐

Дано список чисел. Відсортуйте його за зростанням **без використання методу `.sort()` та функції `sorted()`**.

> 💡 **Підказка:** Найпростіший алгоритм — «бульбашкове сортування»: у вкладеному циклі порівнюємо сусідні елементи і міняємо їх місцями, якщо ліворуч стоїть більший. Повторюємо, поки не пройдемо весь список без жодної заміни.

````python
# Ваш код для завдання 56:
````

````python
numbers = [42, 17, 89, 3, 56, 91, 24]
# Ваш код:


n = len(numbers)

for i in range(n):
    for j in range(0, n - i - 1):        # кожен прохід коротший
        if numbers[j] > numbers[j + 1]:  # сусіди не по порядку?
            numbers[j], numbers[j+1] = numbers[j+1], numbers[j]  # міняємо!

print("Відсортований:", numbers)  # [3, 17, 24, 42, 56, 89, 91]
````

### Завдання 58. Шифр Цезаря ⭐⭐⭐

Запитайте у користувача слово (українською або англійською) і число `shift`.
Зашифруйте слово: кожну літеру замініть на ту, що стоїть на `shift` позицій далі в алфавіті.

**Приклад:** слово `abc`, `shift = 2` → `cde`.

> 💡 **Підказка:** Використайте функції `ord(letter)` (повертає код символу) і `chr(num)` (повертає символ за кодом). Для кожної літери: новий код = ord(letter) + shift. Для простоти можна не обробляти переходи через кінець алфавіту.

````python
# Ваш код для завдання 58:
word  = input("Введіть слово: ")
shift = int(input("Введіть зсув: "))

result = ""
for letter in word:
    result += chr(ord(letter) + shift)   # зсуваємо код символу

print("Зашифровано:", result)
````

### Завдання 59. Гра «Вгадай число» ⭐⭐⭐

Створіть змінну `secret = 42` (загадане число). У циклі `while` запитуйте у користувача число:
- Якщо введене число менше за загадане — пишіть «Більше»
- Якщо більше — «Менше»
- Якщо вгадав — пишіть «Вітаю!» і завершуйте цикл
- Підрахуйте кількість спроб і виведіть її в кінці

> 💡 **Підказка:** Використайте змінну-прапорець `guessed = False` як умову циклу. Коли користувач вгадав — `guessed = True`.

````python
# Ваш код для завдання 59:
secret  = 42
guessed = False      # прапорець
attempts = 0         # лічильник спроб

while not guessed:
    number = int(input("Вгадайте число: "))
    attempts += 1

    if number < secret:
        print("Більше!")
    elif number > secret:
        print("Менше!")
    else:
        print("Вітаю! Ви вгадали!")
        guessed = True

print(f"Кількість спроб: {attempts}")
````

### Завдання 60. Аналіз тексту ⭐⭐⭐

Дано рядок з реченням. Виконайте такий аналіз і виведіть результати:
1. Кількість літер (без пробілів)
2. Кількість слів
3. Кількість голосних літер
4. Найдовше слово

> 💡 **Підказка:** Метод `.split()` розбиває рядок на список слів. Наприклад, `'hello world'.split()` повертає `['hello', 'world']`.

````python
# Ваш код для завдання 60:
````

````python
text = "програмування на python це цікаво і корисно"
# Ваш код:
# 1. Кількість літер без пробілів
letters = 0
for ch in text:
    if ch != " ":
        letters += 1
print("Літер:", letters)

# 2. Кількість слів
words = text.split()
print("Слів:", len(words))

# 3. Голосні
vowels = "аеиіоуяюєї"
count = 0
for ch in text:
    if ch in vowels:
        count += 1
print("Голосних:", count)

# 4. Найдовше слово — патерн з завдання 38!
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print("Найдовше слово:", longest)
````

### Завдання 61. Магазин: статистика продажів ⭐⭐⭐

Дано список продажів за тиждень (по одному числу на день, починаючи з понеділка).
Виведіть:
1. Загальну виручку за тиждень
2. Середню виручку на день
3. Найкращий день (з найбільшою виручкою) — назвою (Понеділок, Вівторок...)
4. Кількість днів, коли виручка була вищою за середню

> 💡 **Підказка:** Створіть список з назвами днів: `days = ["Понеділок", "Вівторок", ...]`. Індекс найкращого дня знайдіть у циклі (або через `sales.index(max(sales))`).

````python
# Ваш код для завдання 61:
````

````python
sales = [3200, 2800, 4100, 3500, 5200, 7800, 6500]
# Ваш код:
days  = ["Понеділок", "Вівторок", "Середа",
         "Четвер", "П'ятниця", "Субота", "Неділя"]

# 1. Загальна виручка
total = sum(sales)
print("Загальна виручка:", total, "грн")

# 2. Середня виручка
average = total / len(sales)
print(f"Середня виручка: {round(average, 2)} грн")

# 3. Найкращий день
best_index = sales.index(max(sales))   # індекс максимального елементу
print("Найкращий день:", days[best_index])

# 4. Днів вище середнього
above = 0
for s in sales:
    if s > average:
        above += 1
print("Днів вище середньої:", above)
````

### Завдання 62. Турнірна таблиця ⭐⭐⭐

Дано два списки: команди та їхні очки (відповідні позиції).
Виведіть турнірну таблицю, ВІДСОРТОВАНУ за спаданням очок, у такому форматі:
```
1. Шахтар — 67 очок
2. Динамо — 64 очки
...
```

> 💡 **Підказка:** Найпростіший спосіб: у циклі знаходьте команду з максимальними очками, виводьте її, потім видаляйте цю пару (команду і очки) з обох списків. Повторюйте, поки списки не спорожніють.

````python
# Ваш код для завдання 62:
````

````python
teams = ["Динамо", "Шахтар", "Зоря", "Дніпро", "Ворскла", "Олександрія"]
points = [64, 67, 52, 48, 39, 45]
# Ваш код:
place = 1

while len(teams) > 0:
    best_index = points.index(max(points))   # індекс лідера

    print(f"{place}. {teams[best_index]} — {points[best_index]} очок")

    teams.pop(best_index)     # видаляємо команду
    points.pop(best_index)    # видаляємо її очки

    place += 1
````
