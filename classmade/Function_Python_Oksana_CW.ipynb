````python
a = 2
b = 3

def sum_num(n1, n2):
    print(n1 + n2)

# sum_num(n1=a, n2=b) # sum_num(n2=b, n1=a)
sum_num(a, b)  # n1=a, n2=b => n1, n2 = a, b
````

````python
my_dict = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five"
}

# Повертає NoneType, без зміни даних
def info(d):
    print(d)

# info(my_dict)
type(info(my_dict))
````

````python
# Повертає NoneType, без зміни даних, виводить результат
def info(d):
    print(d)

type(info(my_dict))
````

````python
# Повертає NoneType, змінює початкові дані, виводить результат
def info(d):
    d[6] = "six"
    print(d)

type(info(my_dict))
````

````python
# Повертає NoneType, змінює початкові дані
def info(d):
    d[6] = "six"

print(f"my_dict: {my_dict}")
type(info(my_dict))
````
